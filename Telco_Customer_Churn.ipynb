```python
df1 = df.drop(["customerID", "gender"], axis=1)
```

```python
df1["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df1["TotalCharges"] = df1["TotalCharges"].fillna(df1["TotalCharges"].median())
```

```python
df1["Churn"] = df1["Churn"].map({
    "No": 0,
    "Yes": 1
    })
```

```python
df_encoded = pd.get_dummies(df1, drop_first=True, dtype=int)
df_encoded.head()
```

```python
x = df_encoded.drop(columns="Churn")
x
```

```python
y = df_encoded["Churn"]
y
```

```python
x.dtypes.value_counts()   # should show only int64 and float64, no 'object'
x.shape                   # about (7043, 28)
```

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.30, random_state=0, stratify=y)
```

```python
from imblearn.over_sampling import SMOTE

sm = SMOTE()

xtrain1, ytrain1 = sm.fit_resample(xtrain, ytrain)
```

```python
ytrain1.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(xtrain1)
xtrain1 = scaler.transform(xtrain1)

xtrain1
```

```python
xtest = scaler.transform(xtest)
xtest
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(n_estimators = 100, criterion = "entropy", max_depth = 8, random_state=0)
```

```python
rfc.fit(xtrain1, ytrain1)
```

```python
ypred = rfc.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(ytest, ypred)
```

```python
from xgboost import XGBClassifier
xgb = XGBClassifier(random_state=0)
```

```python
xgb.fit(xtrain1, ytrain1)
ypred = xgb.predict(xtest)
```

```python
accuracy_score(ytest, ypred)
```
